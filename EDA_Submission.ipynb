## ***4. Data Vizualization, Storytelling & Experimenting with charts : Understand the relationships between variables***

#### Chart - 1

````python
# Chart - 1 visualization code
plt.figure(figsize=(8,5))

sns.countplot(x='CSAT Score', data=df, palette='viridis')

plt.title('Distribution of CSAT Scores')
plt.xlabel('CSAT Score')
plt.ylabel('Number of Customers')

plt.show()
````

##### 1. Why did you pick the specific chart?

A Count Plot was chosen because the CSAT Score is a categorical variable with ratings ranging from 1 to 5. This chart effectively displays the frequency distribution of each satisfaction score, making it easy to identify customer satisfaction trends and detect any imbalance in the ratings.

##### 2. What is/are the insight(s) found from the chart?

The majority of customers have given a CSAT Score of 5, indicating a high level of customer satisfaction.
Scores of 4 and 1 are the next most frequent ratings, while 2 and 3 occur much less often.
The distribution is highly imbalanced, with positive feedback significantly outweighing neutral and moderately dissatisfied responses.
Although overall satisfaction is high, the noticeable number of 1-star ratings suggests that a segment of customers experienced poor service and should be investigated further.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. The chart indicates that most customers are highly satisfied with the support services, which reflects positively on the organization's customer service quality and can contribute to improved customer retention and brand reputation.

However, the presence of a considerable number of 1-star ratings highlights areas where customer expectations were not met. Identifying the reasons behind these low ratings—such as specific issue categories, support agents, response delays, or communication quality—can help the organization implement targeted improvements. Addressing these issues can reduce customer dissatisfaction, improve future CSAT scores, and minimize the risk of customer churn.

#### Chart - 14 - Correlation Heatmap

##### 1. Why did you pick the specific chart?

A Correlation Heatmap was chosen to examine the strength and direction of relationships between the numerical variables in the dataset. It provides a quick visual summary of positive, negative, and weak correlations.

##### 2. What is/are the insight(s) found from the chart?

Item Price and Connected Handling Time have a moderate positive correlation (0.65), suggesting that higher-priced items may require longer handling times.
CSAT Score has a very weak positive correlation (0.048) with Connected Handling Time, indicating that handling time has little influence on customer satisfaction.
CSAT Score shows a weak negative correlation (-0.12) with Item Price, suggesting that item price has minimal impact on customer satisfaction.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. The heatmap indicates that numerical variables have little influence on CSAT Score, suggesting that customer satisfaction is more likely driven by service quality, issue type, and customer experience rather than item price or handling time. #### Chart - 15 - Pair Plot

````python
# Pair Plot visualization code
# Select numerical columns
pairplot_df = df[['CSAT Score', 'Item_price', 'connected_handling_time']].copy()

# Convert columns to numeric
pairplot_df['Item_price'] = pd.to_numeric(pairplot_df['Item_price'], errors='coerce')
pairplot_df['connected_handling_time'] = pd.to_numeric(
    pairplot_df['connected_handling_time'],
    errors='coerce'
)

# Remove missing values
pairplot_df = pairplot_df.dropna()

# Pair Plot
sns.pairplot(pairplot_df)

plt.show()
````

##### 1. Why did you pick the specific chart?

A Pair Plot was selected to examine the pairwise relationships and distributions among the numerical variables in the dataset. It helps identify correlations, trends, clusters, and potential outliers by displaying scatter plots between numerical features along with their individual distributions.

##### 2. What is/are the insight(s) found from the chart?

The Pair Plot allows visual inspection of relationships between CSAT Score, Item Price, and Connected Handling Time.
Due to the large amount of missing values in Item_price and connected_handling_time, the number of observations available for analysis is limited.
No strong linear relationship may be observed between the available numerical variables, suggesting that customer satisfaction is likely influenced more by categorical factors such as issue category, support channel, or agent characteristics.

##### 3. Will the gained insights help creating a positive business impact?
Are there any insights that lead to negative growth? Justify with specific reason.

Yes. Although the numerical relationships are weak, the Pair Plot confirms that numerical variables alone are insufficient to explain customer satisfaction. This insight directs the analysis toward categorical features, which are more likely to influence CSAT scores. Understanding the true drivers of customer satisfaction enables the organization to focus improvement efforts on operational processes rather than relying solely on numerical metrics.
